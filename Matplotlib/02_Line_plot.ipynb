```python
import matplotlib.pyplot as plt
courses = ['Computer Science', 'Data Science', 'AI & ML', 'Cybersecurity', 'Web Development']
students_enrolled = [120, 95, 110, 70, 85]
students_left = [40, 30, 10, 10, 25]
plt.plot(courses,students_enrolled,color="blue",marker="o",label="Students Enrolled")
plt.plot(courses,students_left,color="green",marker="*",alpha=0.4,label="Students Left")
plt.legend()
plt.show()
```
