```python
import pandas as pd
data=pd.read_csv("C:/Users/Lenovo/Downloads/myExpenses1.csv")
df=pd.DataFrame(data)
print(data)
```

```python
expense_by_item = df.groupby("Item")["Amount"].sum()
plt.boxplot(expense_by_item)
plt.show()
```

```python
plt.figure(figsize=(8,6))
df.boxplot(column="Amount", by="day")

plt.title("Expense Distribution by Day")
plt.suptitle("")
plt.xlabel("Day")
plt.ylabel("Amount Spent")
plt.show()
```
