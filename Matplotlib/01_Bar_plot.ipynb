```python
import matplotlib.pyplot as plt
courses = ['Computer Science', 'Data Science', 'AI & ML', 'Cybersecurity', 'Web Development']
students_enrolled = [120, 95, 110, 70, 85]
plt.figure(figsize=(10, 6),layout="tight")
plt.bar(courses, students_enrolled, color='skyblue', edgecolor='black')
plt.xlabel('Courses')
plt.ylabel('Number of Students Enrolled')
plt.title('Student Enrollment in Different Courses')
plt.grid(axis='y', linestyle='--', alpha=0.51)
plt.show()
```
