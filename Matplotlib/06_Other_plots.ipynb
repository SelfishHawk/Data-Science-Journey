```python
#HISTOGRAM
import matplotlib.pyplot as plt
x=[12,15,26,37,23,36,25,28,30]
plt.hist(x, len(x),edgecolor="black",color="pink")
plt.show
```

```python
#Violin plot
plt.violinplot(x,showmedians=True)
plt.show()
#Greater the buldge sideways means more the values exist there, also it shows  how data is related to the medians
```

```python
#Stem plot
plt.stem(x,linefmt="--",markerfmt="D",bottom=10)
plt.show()
```

```python
#Stack plot
days=[1,3,4,5,7]

NOP1=[50,10,30,50,70]
NOP2=[10,30,40,60,90]
NOP3=[2,100,200,30,500]
plt.stackplot(days,NOP1,NOP2,NOP3,labels=["week1","week2","week3"])
plt.legend()
plt.show()
```

```python
x=["day1","day2","day3","day4","day5"]
y=[40,20,30,50,40]
plt.step(x,y,where="mid",marker="D") #where-"pre/post/mid"
plt.show()
```

```python
#More about Graphs
#Legends
x=[1,3,5,7,9,11,13,15]
y=[2,4,6,8,10,12,14,16]
y1=[5,10,15,20,25,30,40,50]
y2=[10,20,30,40,50,60,70,80]
plt.plot(x,y,label="x & y")
plt.plot(x,y1,label="x & y1")
plt.plot(x,y2,label="x & y2")
plt.plot(y1,y2,label="y2 & y1")
plt.legend(loc=1,ncols=20,bbox_to_anchor=(1,1.1),labelspacing=5)
plt.show()
```

```python
#Sub plot
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.subplot(1,2,1)
plt.title("Fig 1")
plt.plot(x,y)

x=[6,7,8,9,10]
y=[1,3,7,9,4]
plt.subplot(1,2,2)
plt.title("Fig 2")
plt.plot(x,y)
plt.suptitle("Figure")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Example pie chart
sizes = [30, 25, 20, 25]
labels = ["Chai", "Juice", "Coffee", "Snacks"]

plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=140)
plt.title("Expense Distribution")

# Save the figure
plt.savefig("expenses_pie.png", dpi=300, bbox_inches="tight")

plt.show()
```
