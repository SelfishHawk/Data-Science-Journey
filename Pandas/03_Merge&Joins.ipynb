```python
import pandas as pd
data1={
    "EEID":["E01","E02","E03","E04","E05","E06"],
    "Names":["Vishal","Raj","Harish","Harsh","Shyam","Prag"],
    "Age":[32,35,66,44,19,20]
}
data2={
    "EEID":["E01","E02","E03","E04","E05","E06"],
   "Salary":[25000,65000,99000,56000,20000,30000] 
}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
print(df1)
print()
print(df2)
```

```python
print(pd.merge(df1,df2,on="EEID"))
```

```python
data3={
    "EEID":["E01","E02","E03","E04","E05","E06"],
    "Names":["Vishal","Raj","Harish","Harsh","Shyam","Prag"],
    "Age":[32,35,66,76,19,20]
}
data4={
    "EEID":["E01","E02","E07","E04","E08","E06"],
   "Salary":[25000,65000,99000,56000,20000,30000] 
}
df3=pd.DataFrame(data3)
df4=pd.DataFrame(data4)
print(df3)
print()
print(df4)
```

```python
print(pd.merge(df3,df4,on="EEID"))
print()
print(pd.merge(df3,df4,on="EEID",how="left"))
print()
print(pd.merge(df3,df4,on="EEID",how="right"))
```

```python
dict={
    "Fruit":["Mango","Apple","Orange","Guava"],
    "Quantity":["25","20","30","10"],
    "Price":["45","40","20","50"]
}
d1=pd.DataFrame(dict)
d2=d1.copy()
d2.loc[0,"Price"]=48
d2.loc[1,"Price"]=35
d2.loc[3,"Price"]=55
d2.loc[0,"Quantity"]=20
d2.loc[1,"Quantity"]=30
d2.loc[3,"Quantity"]=15
print(d1)
print()
print(d2)
```

```python
print(d1.compare(d2))
```

```python
print(d1.compare(d2,align_axis=0))
```

```python
print(d1.compare(d2,keep_equal=True))
```

```python
print(d1.compare(d2,keep_shape=True))
#Keep shape default value is false
```
