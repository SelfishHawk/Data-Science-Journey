```python
#Defining Class
class Student:
    name="Karan"
s1=Student()
print(s1.name)
```

```python
#Constructor
class Cars:
    def __init__ (self,br,clr):
        print("Constructor called")
        self.brand=br
        self.colour=clr
    def info(self):
        print(f"Car is of {self.brand} ,available in {self.colour} colour")
GTR=Cars("Nissan","White")
M14=Cars("Forgot","Black")
Gemera=Cars("Koinsegg","Yellow")
GTR.info()
Gemera.info()
```

```python
#Attributes
class Student:
    College_name="ABC Institue" # Class Attributes
    def __init__(self,name,marks):
        self.name=name# Instance Attributes
        self.marks=marks
s1=Student("Karan",98)
s2=Student("Arjun",75)
print(s2.name,s2.College_name)
print(Student.College_name)
```

```python
#Methods
class Marks:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def info(self):
        print(f"{self.name} got {self.marks}")
    @staticmethod #Static methods
    def hello():
        print("Hello")
s1=Marks("Karan",98)
s1.info()
s2=Marks("Rehan",88)
s2.info()
s2.hello()
```
