```python
#Inheritance 
class Car:
    @staticmethod
    def start():
        print("Car started....")
    @staticmethod
    def stop():
        print("Car Stopped....")
class BMW(Car):
    def __init__(self,car_name,type):
        self.car_name=car_name
        self.type=type
        
    def show_type (self):
        print("type of the car is ",self.type)
car1=BMW("S_Class","Diesel")
car1.start()
car1.show_type()
```

```python
#Property
class Num:
    def __init__(self,math,phy,chem):
        self.math=math
        self.phy=phy
        self.chem=chem
    @property
    def Percentage(self):
        return str((self.phy+self.chem+self.math)/3)+"%"
stu1=Num(98,95,75)
print(stu1.Percentage)
stu1.chem=98
print(stu1.Percentage)
```

```python
#Polymorphism
#Character Overloading
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img=img
    def ShowNum(self):
        print(f"{self.real} i + {self.img} j")
    def __add__(self,num2):
        NewReal=self.real + num2.real
        NewImg=self.img + num2.img
        return Complex(NewReal,NewImg)
    def __sub__(self,num2):
        NewReal=self.real - num2.real
        NewImg=self.img - num2.img
        return Complex(NewReal,NewImg)
num1=Complex(4,5)
num1.ShowNum()
num2=Complex(1,3)
num2.ShowNum()
num3=num1+num2
num3.ShowNum()
num4=num3 - num2
num4.ShowNum()
```

```python
#Practice Question
class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*(self.radius **2)
    def perimeter(self):
        return 2*3.14*(self.radius)


circle1=Circle(10)
print(circle1.area())
print(circle1.perimeter())
```
